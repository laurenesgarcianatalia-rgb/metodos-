MÉTODO DE BISECCIÓN

Queremos resolver la ecuacion $x^3+4x^2-10=0$ en el intervalo $[1,2]$

EL *método de bisección* consiste en dividir a la mitad em intervalo en cada iteración. El punto medio Pn es la solución aproximada

```python
from math import *
import numpy as np
import matplotlib.pyplot as plt
```

```python
def f(x):
  return x**3 + 4*x**2 - 10.0
```

```python
f(1)
```

```python
x = np.linspace(0,3) #genera una reticula o una particion
plt.plot(x, f(x), label="f(x)")
plt.title("Gráfica de la función $f(x) = x^3 + 4x^2 - 10 = 0$") #titulo
plt.xlabel("$x$")
plt.ylabel("$f(x)$")
plt.axhline(y=0, color='red', linestyle='--', label='y = 0 (Eje X)') # Dibuja la línea del eje X
plt.legend() #muestra la informacion de las rayas de colores
plt.grid(True) #activa la cuadricula
plt.show()
```

```python
from matplotlib import matplotlib_fname
#ingreso datos de entrada para los diferentes metodos a trabajar
a = 1
b = 2
#guardar valores iniciales
a0 = a
b0 = b
#guardar valores inicames del error y del numero de iteraciones
tol = 0.00001 #float(input("Ingrese el valor de la tolerancia: "))
max = 100 #float(input("Ingrese el numero macimo de iteraciones: "))
error = 100
niter = 0

#Metodo de bisección

#evaluo primer balor medio

m= a + (b-a)/2

#Evaluacion de la funcion en los puntos a, b y m
fa = f(a)
fb = f(b)
fm = f(m)

print("# iter\t\t a \t\t f(a) \t\t b \t\t f(b) \t\t m \t\t f(m) \t\t error")
print("{0} \t\t {1:6.4f} \t {2:6.4f} \t {3:6.4f} \t {4:6.4f} \t {5:6.4f} \t {6:6.4f} \t {7:6.4f}".format(niter,a0,fa,b0,fb,m,fm,error))

#ciclo iterativo
while error > tol and niter < max:
  m = a + (b-a) /2
  if np.sign(fa) == np.sign(fm):
    a = m
    fa = f(a)
  else:
    b = m
    fb = f(b)

  m = a + (b-a) /2
  fm = f(m)
  error = abs(b-a)
  niter +=1

print("{0} \t\t {1:6.6f} \t {2:6.6f} \t {3:6.6f} \t {4:6.6f} \t {5:6.6f} \t {6:6.6f} \t {7:6.6f}".format(niter, a, fa, b, fb, m, fm, error ))

print("La raíz de la función dada en el intervalo [{0:6.4f},{1:6.4f}] es {2:6.7f}".format(a0,b0,m))
```
